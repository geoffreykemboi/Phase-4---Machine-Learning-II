```python
# remove hashtags#, urls, @ clean up, convert data to lower case.

import re

# urls
pat_1 = r'https:/S+|www\.\s+'
# hashtages and mentions
pat_2 = r'#\w+|@\w+'
# punctuations and hashtags
pat_3 = r'[^a-z\s]'
```

```python
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords


STOPWORDS = set(stopwords.words('english'))

def cleaner(text):
    """ 
    lowercase, then remove the defined patterns from the tweets
    """
    if not isinstance(text, str):
        return 'This is not a valid tweet(text)!!!'
    
    text = text.lower()

    #remove url
    text = re.sub(pat_1, '',text)

    #remove mentions and hashtags
    text = re.sub(pat_2, '',text)

    #remove punctuations and random numbers
    text = re.sub(pat_3, '',text)


    tokens = word_tokenize(text)

    #list comprehension on tokens to remove stop words
    tokens = [token for token in tokens if token not in STOPWORDS and len(token) > 2]

    return " ".join(tokens)
```

```python
import re
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

STOPWORDS = set(stopwords.words('english'))

def cleaner(text):
    """ 
    lowercase, then remove the defined patterns from the tweets
    """
    if not isinstance(text, str):
        return 'This is not a valid tweet(text)!!!'
    
    text = text.lower()

    # remove url
    text = re.sub(pat_1, '', text)

    # remove mentions and hashtags
    text = re.sub(pat_2, '', text)

    # remove punctuations and random numbers
    text = re.sub(pat_3, '', text)

    tokens = word_tokenize(text)

    # remove stopwords and short words
    tokens = [token for token in tokens if token not in STOPWORDS and len(token) > 2]

    return " ".join(tokens)
```
