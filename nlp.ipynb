### Natural Language Processing.
NLP refers to analytics tasks that deal with human language in the form of texts or speech.

Unlike normal csv files or data read in from databases, for NLP tasks, you need to preprocess the data before its usable in analytics. 

#### Regular Expressions
Regular Expressions or *Regex* is an essential part of NLP used to quickly find patterns and filter through documents.

After processing the data and getting it in a working format, the next huddle you'll face is ambiguity. Most sentences, for example, in English language use and reuse the same words. Thus finessing when feature engineering is required.

Here are the techniques used in feature engineering for text data:
- Removing stop words
- Stemming 
- Lemmatization

Moreover, we'll cover **context-free grammars** and **part-of-speech** tagging. By definition, context-free grammars refer to bits of text that are grammatically correct but feel like complete nonsense when considering the same bit of text on the semantic level. 

POS tagging refers to the act of helping a computer understand how to intepret a sentence. 

Finally we'll finish by explaining the general process to set text data up for classification problems.

### Tokenization.


```python
s1 = 'The cat stole the stew from the pot.'
s2 = 'The rains have been heavy.'
s3 = 'It was raining cats and dogs.'
s4 = 'Dogs bark when stealing.'
```

```python
samples = [s1, s2, s3, s4]

lower_cased_samples = []

for sample in samples: 
    lower_cased_samples.append(sample.lower())
    print(sample.lower().split())
```

### Vectorization.
CountVectorization.

```python
sentence1 = 'These floods in Nairobi were avoidable.'
sentence2 = 'The weather seems to be unpredictable.'
sentence3 = 'Early warning is crucial in disaster avoidance.'

paragraph = [sentence1, sentence2, sentence3]

lowercased_paragraph = []

for sentence in paragraph:
    lowercased_paragraph.append(sentence.lower())

lowercased_paragraph[0]
```

```python
from sklearn.feature_extraction.text import CountVectorizer

#instantiate
count_vectorizer = CountVectorizer()
```

```python
count_vectorizer.fit_transform(lowercased_paragraph)

count_vectorizer.vocabulary_
```

```python
count_vectorizer.fit_transform(lower_cased_samples)

count_vectorizer.vocabulary_
```

```python
count_vectorizer = CountVectorizer(stop_words='english')

count_vectorizer.fit_transform(lower_cased_samples)

count_vectorizer.vocabulary_
```
